## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success Metric**

The primary success metric for this project is **ROC-AUC (Receiver Operating Characteristic – Area Under the Curve)**. ROC-AUC measures how well the model distinguishes between declining and non-declining content across different decision thresholds. A value of **0.5** indicates random guessing, while **1.0** represents perfect classification. For this project, a **ROC-AUC of 0.80 or higher** would indicate that the model has good predictive performance and can provide useful decision support for prioritizing content updates.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML Beats a Fixed Rule**

A fixed rule, such as selecting pages with a low CTR or poor search position, cannot capture the complex relationships between multiple performance signals. Content performance depends on the interaction of factors such as clicks, impressions, CTR, average position, engagement metrics, content characteristics, and historical trends. These relationships are often nonlinear and vary across different types of content and clients. A machine learning model can learn these patterns from historical data and combine many features to make more accurate predictions than a single hand-written rule, making it a better tool for supporting content prioritization decisions.
