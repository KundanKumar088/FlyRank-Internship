### 1. Research Question

**Research Question**

Can a machine learning model rank website pages more effectively than FlyRank's existing rule-based flags for identifying which pages should be prioritized for manual content review?

**Decision It Supports**

The research supports the decision of **which pages a content team should review first** when time and resources are limited. Instead of relying only on fixed rule-based flags, the model provides a ranked list of pages based on historical performance patterns, helping teams prioritize manual review more efficiently. The model is intended as a **decision-support tool**, not as a replacement for human judgment or a predictor of Google's search algorithm.


## 2. Data

### Data

The study used the **public-safe FlyRank internship dataset**, which is a pseudonymized release derived from historical website performance data. The data originates from **Google Search Console** and **Google Analytics**, is aggregated in **BigQuery**, and provided to interns without any private client information.

The analysis used the tables supplied in the internship release, containing page-level historical metrics such as impressions, clicks, click-through rate (CTR), average position, engagement, and other aggregated performance features. The dataset covers the historical date window included in the release, with rolling feature windows (for example, recent 7-, 30-, or 90-day aggregates where provided).

To ensure reliable model training, records with missing key fields, duplicate entries, or insufficient historical information were excluded. Product-generated outputs such as **health scores** and **rule-based flags** were also excluded from the feature set because they represent existing decisions and would introduce data leakage. All data used in the analysis is public-safe and contains no client names, raw search queries, personally identifiable information, or confidential business data.


## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

###  Methodology

The objective was to determine whether a machine learning model could prioritize pages for manual review more effectively than the existing rule-based approach. The analysis assumes that historical page performance contains useful signals for identifying pages that deserve attention, while recognizing that these patterns are **associational** rather than causal.

The model was trained using historical performance features provided in the internship dataset, including metrics such as impressions, clicks, click-through rate (CTR), average position, engagement metrics, and rolling historical aggregates (where available). Features that represented existing product decisions, such as **health scores** and **rule-based flags**, were excluded because they would introduce data leakage.

The target label was defined using the internship dataset's provided review label, indicating whether a page should be prioritized for manual review. This label was used only as the prediction target and was never included as an input feature.

As a baseline, the existing rule-based prioritization (product flags) and a simple baseline model were used for comparison. The machine learning models were then evaluated against the same target using identical evaluation metrics.

To obtain an honest estimate of performance, the data was evaluated using a **grouped validation split**, ensuring that pages from the same client did not appear in both the training and testing sets. This prevents the model from memorizing client-specific patterns and provides a more realistic assessment of how well it generalizes to unseen data.

A leakage audit was performed before training. Product outputs (such as health scores and flags), future-looking variables, duplicate identifiers, and any features derived from the target label were removed from the training data. The feature set was also reviewed to confirm that all inputs would be available at prediction time. This validation process helps ensure that the reported performance reflects genuine predictive ability rather than information leakage.

## 5. Limitations

*What this work cannot claim.*

### Limitations

This study has several limitations that should be considered when interpreting the results. First, the analysis is based on a **single historical, pseudonymized FlyRank internship dataset**, so the findings may not generalize to other websites, industries, or future time periods. Second, the work uses **observational data**, which means the identified relationships are associations rather than evidence of causation. Therefore, the results should not be interpreted as showing that any specific content change will improve search rankings or traffic.

Although grouped client-level validation and leakage checks were used to obtain an honest estimate of model performance, real-world performance may change as search behavior, user preferences, and search engine algorithms evolve. Periodic retraining and evaluation on newer data would be necessary to maintain performance.

Finally, the model is designed as a **decision-support tool** to help prioritize pages for manual review. It does not replace editorial expertise, nor does it predict or model Google's search algorithm. Final content decisions should always include human judgment alongside the model's recommendations.
