## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review + No-Go List

Before acting on a recommendation, a reviewer should check that the page is still relevant, the data is current, and there is a clear reason for making changes. They should verify the content quality, search intent, recent performance trends, and any business priorities that the model cannot see. If new information has become available since the data was collected, the recommendation should be reconsidered.

The following decisions should **never be fully automated** based only on the model's output:

* Publishing, rewriting, or deleting content without human review.
* Making major SEO changes solely because a page received a high model score.
* Treating the model's ranking as proof that an action will improve performance.
* Applying the same recommendation to every page without considering its context or business goals.

The model should be used to **prioritize pages for review**, while the final decision remains with a human reviewer.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

###  Monitoring / Retrain Triggers

The model should be monitored regularly to ensure its recommendations remain reliable. If Precision@50 or other validation metrics decline over time, it may indicate that the model is no longer capturing current patterns. Large changes in website content, user behavior, or search engine updates are also signs that retraining may be needed.

The recommendations should also be reviewed if the distribution of key features changes significantly or if reviewers consistently find that high-ranked pages are poor candidates for action. In these situations, the model should be retrained using newer data and evaluated again with the same grouped or time-aware validation before being used for decision support.
