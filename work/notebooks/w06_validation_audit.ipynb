## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

### 3. Leakage audit

I reviewed my final feature set before training the model to make sure there was no information leakage.

* **Label-derived features:** I confirmed that `trend_pct` and `trend_direction` were **not included** because the target (`is_declining_label`) is derived from them. Including these columns would leak the answer.

* **Future information:** All features were taken from data that would have been available **before the prediction point**. I did not use any metrics that overlap with the label window.

* **Product flags:** I excluded existing system scores, product flags, or decision-based columns from the feature set. These can be useful as baselines but should not be model inputs.

* **IDs:** `content_id` and `client_id` were used only for identification and grouped validation. They were **not used as model features**.

* **Missing values:** Instead of blindly filling missing values with zero, I handled missing data carefully and used missing-value indicators where appropriate to avoid introducing artificial signals.

**Conclusion:**
After this audit, I found no evidence of label leakage, future leakage, or decision leakage in my final feature set. The model was trained only on information that would realistically be available at prediction time.
